```python
import whisper
import json
import os

model = whisper.load_model("large-v2")

result = model.transcribe(
    "/content/RAG/audios/10_YTDown.com_YouTube_Installing-VS-Code-_-How-Websites-Work-S_Media_tVzUXW6siu0_002_720p tutorial.mp3",
    language="hi",
    task="translate",
    word_timestamps=False
)
chunks= []
for segment in result["segments"]:
  chunks.append({"start": segment["start"], "end": segment["end"], "text": segment["text"],"number": segment["nummber"],"tittle":segment["tittle"]})

print(chunks)

chunks_with_metadata={
    "chunks":chunks,
    "text":result["text"]
}
os.makedir("json",exist_ok=True)
with open("output.json","w") as f:
  json.dump(chunks_with_metadata, f,indent=4)

```
